# Real token embeddings

Instead of hand-picked 3-dim vectors, use GPT-2's learned token embeddings (768-dim) for the same sentence.

```python
import tiktoken
import torch
from transformers import GPT2Model

tokenizer = tiktoken.get_encoding("gpt2")
model = GPT2Model.from_pretrained("gpt2")
```

```python
sentence = "Your journey starts with one step"
token_ids = tokenizer.encode(sentence)
words = [tokenizer.decode([t]).strip() for t in token_ids]

# GPT-2's learned token embedding table (wte), one 768-dim vector per token
with torch.no_grad():
    inputs = model.wte(torch.tensor(token_ids))   # shape (6, 768)

print(words)
print(inputs.shape)
```
